### Q.2 Questions on Basic NumPy Array:
a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])

b) Find the most frequent value and their indice(s) in the following arrays:
i. x = np.array([1,2,3,4,5,1,2,1,1,1])
ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

```python
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print(arr)
print(reversed_arr)

from collections import Counter

def most_frequent_value_and_indices(arr):
    counts = Counter(arr)
    if not counts:
        return "Array is empty", []
    most_common_value = counts.most_common(1)[0][0]
    indices = np.where(arr == most_common_value)[0]
    return most_common_value, indices

x = np.array([1,2,3,4,5,1,2,1,1,1])
most_freq_x, indices_x = most_frequent_value_and_indices(x)
print(f"\nArray x: {x}")
print(f"Most frequent value in x: {most_freq_x}")
print(f"Indices of most frequent value in x: {list(indices_x)}")

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
most_freq_y, indices_y = most_frequent_value_and_indices(y)
print(f"\nArray y: {y}")
print(f"Most frequent value in y: {most_freq_y}")
print(f"Indices of most frequent value in y: {list(indices_y)}")
```

### Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?

```python
import numpy as np

# Create 1-D NumPy array named after you with 25 evenly spaced numbers from 10 to 100
agrim_gupta = np.linspace(10, 100, 25)
print("Array (agrim_gupta):\n", agrim_gupta)

# Print attributes of the array
print(f"\nDimensions: {agrim_gupta.ndim}")
print(f"Shape: {agrim_gupta.shape}")
print(f"Total elements: {agrim_gupta.size}")
print(f"Data type: {agrim_gupta.dtype}")
print(f"Total number of bytes: {agrim_gupta.nbytes} bytes")

# Transposing a 1D array using reshape() to convert it into a 2D column vector first
# A true mathematical transpose swaps dimensions, changing a row format to a column format.
print("\n--- Transposing via reshape() ---")
reshaped_column = agrim_gupta.reshape(-1, 1)
print(f"Reshaped to 2D column vector (shape {reshaped_column.shape}):\n{reshaped_column}")

transposed_via_reshape = reshaped_column.T
print(f"Transpose of that column vector using .T (shape {transposed_via_reshape.shape}):\n{transposed_via_reshape}")

# Demonstration of using .T directly on the original 1D array
print("\n--- Transposing via .T directly ---")
transposed_by_T_1d = agrim_gupta.T
print(f"Direct .T on 1D array (shape {transposed_by_T_1d.shape}):\n{transposed_by_T_1d}")

# Explanation answer to the question: "Can we do the same with T attribute?"
print("\n[ANSWER]: No, applying the .T attribute directly to a 1D array does not change its layout.")
print("A 1D array only has one dimension, so swapping axes results in the exact same 1D shape.")
print(f"Is original 1D array identical to its direct .T: {np.array_equal(agrim_gupta, transposed_by_T_1d)}")
```
