```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Original temperature readings (Celsius):")
display(temperature)
```

### a. Add 2°C to every reading using vectorization.

```python
corrected_temperature = temperature + 2
print("Temperature readings after adding 2°C:")
display(corrected_temperature)
```

### b. Convert all temperatures from Celsius to Fahrenheit.

```python
fahrenheit_temperature = (corrected_temperature * 9/5) + 32
print("Temperature readings converted to Fahrenheit:")
display(fahrenheit_temperature)
```

### c. Identify all readings greater than 32°C using Boolean indexing.

```python
readings_greater_than_32C = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:")
display(readings_greater_than_32C)
```

### d. Count how many readings exceed 32°C.

```python
count_greater_than_32C = np.sum(corrected_temperature > 32)
print(f"Number of readings exceeding 32°C: {count_greater_than_32C}")
```

### e. Explain how vectorization and Boolean indexing can make data processing more efficient than explicitly iterating through every value.

**Vectorization** is the process of performing operations on entire arrays or matrices at once, rather than element by element. In Python, libraries like NumPy implement these operations in highly optimized, pre-compiled C or Fortran code. This significantly reduces the overhead associated with Python's interpretative nature and explicit loops, leading to much faster execution times, especially for large datasets.

**Boolean indexing** allows you to select subsets of an array based on a boolean condition applied across the entire array. Instead of writing a loop to check each element and then conditionally add it to a new list, Boolean indexing performs this filtering operation efficiently at a low level. It directly returns a new array containing only the elements where the condition was true.

**Efficiency Benefits:**
1.  **Speed:** Both methods leverage optimized C/Fortran implementations, which are orders of magnitude faster than Python loops for numerical operations.
2.  **Readability:** The code becomes more concise and easier to understand as it reflects the mathematical operations more directly without verbose looping constructs.
3.  **Memory Usage (sometimes):** While not always the case, vectorized operations can sometimes be more memory-efficient by avoiding the creation of intermediate lists or variables within loops.
4.  **Simplicity:** They reduce the complexity of the code, making it less prone to errors.

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
print("Original steps matrix:")
display(steps)
```

### a. Calculate the total steps recorded.

```python
total_steps = np.sum(steps)
print(f"Total steps recorded: {total_steps}")
```

### b. Calculate the mean number of steps.

```python
mean_steps = np.mean(steps)
print(f"Mean number of steps: {mean_steps:.2f}")
```

### c. Find the maximum and minimum values.

```python
max_steps = np.max(steps)
min_steps = np.min(steps)
print(f"Maximum steps: {max_steps}")
print(f"Minimum steps: {min_steps}")
```

### d. Calculate the total steps for each day using axis=0.

```python
total_steps_per_day = np.sum(steps, axis=0)
print("Total steps for each day:")
display(total_steps_per_day)
```

### e. Calculate the total steps for each user using axis=1.

```python
total_steps_per_user = np.sum(steps, axis=1)
print("Total steps for each user:")
display(total_steps_per_user)
```

### f. Find the user/day position containing the maximum number of steps using argmax().

```python
max_position_flat = np.argmax(steps)
max_position_2d = np.unravel_index(max_position_flat, steps.shape)
print(f"Position of the maximum steps (row, column): {max_position_2d}")
print(f"The maximum number of steps ({steps[max_position_2d]}) was recorded by User {max_position_2d[0]} on Day {max_position_2d[1]}.")
```

## Q4. Cognitive-Assistive System: Ordinary Least Squares (OLS) Regression

### Initializing `y` and `X`

```python
y = np.array([40, 65, 30, 85])
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

print("Assistance scores (y):")
display(y)
print("User features (X):")
display(X)
```

### a. Display the shape and dimensions of X.

```python
print(f"Shape of X: {X.shape}")
print(f"Dimensions of X (ndim): {X.ndim}")
```

### b. Calculate: X.T and explain what the transpose represents.

```python
X_T = X.T
print("Transpose of X (X.T):")
display(X_T)
```

#### Explanation of Transpose (X.T):

The transpose of a matrix `X` (denoted as `X.T`) is a new matrix where the rows of `X` become the columns of `X.T`, and the columns of `X` become the rows of `X.T`. Essentially, the element at row `i` and column `j` in `X` is moved to row `j` and column `i` in `X.T`.

In the context of the `X` matrix representing user features (rows are users, columns are features):
*   The original `X` matrix has rows representing individual users and columns representing `Sleep hours`, `Activity level`, and `Stress level`.
*   Its transpose, `X.T`, rearranges this so that the rows now represent the features (`Sleep hours`, `Activity level`, `Stress level`), and the columns represent the individual users. This transformation is often used in linear algebra calculations, such as matrix multiplication for calculating covariances or in the Ordinary Least Squares formula.

### c. Calculate the matrix product: X.T @ X

```python
XT_X = X.T @ X
print("Matrix product (X.T @ X):")
display(XT_X)
```

### d. Calculate the matrix inverse using: `np.linalg.inv()` where appropriate.

```python
XT_X_inverse = np.linalg.inv(XT_X)
print("Inverse of (X.T @ X):")
display(XT_X_inverse)
```

### e. Implement the Ordinary Least Squares equation: $\beta = (X^T X)^{-1} X^T y$ using NumPy.

```python
beta_coefficients = XT_X_inverse @ X.T @ y
print("OLS coefficients (beta):")
display(beta_coefficients)
```

### f. Explain what the resulting coefficients represent in the context of the three user features.

#### Explanation of Coefficients:

The coefficients represent the estimated change in the predicted assistance score for a one-unit increase in each feature, while the other features remain constant.

- **Sleep hours:** The coefficient is approximately `-0.1592`.
- **Activity level:** The coefficient is approximately `0.1408`.
- **Stress level:** The coefficient is approximately `10.0630`.

The model uses the equation `Assistance Score = (β₁ × Sleep Hours) + (β₂ × Activity Level) + (β₃ × Stress Level)`. No intercept is included because the given design matrix `X` does not contain a column of ones.


### g. A new user has: `new_user = np.array([5, 40, 7])` Use your calculated coefficients to obtain a predicted assistance score.

```python
new_user = np.array([5, 40, 7])
predicted_score = np.dot(new_user, beta_coefficients)

print(f"New user features: {new_user}")
print(f"Predicted assistance score for the new user: {predicted_score:.2f}")
```
